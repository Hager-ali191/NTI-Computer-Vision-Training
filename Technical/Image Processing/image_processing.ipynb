```python
from skimage import transform
import numpy as np
```

```python
transf = transform.SimilarityTransform(np.array([
    [1, 0, 100],
    [0, 1, 100],
    [0, 0, 1]
]))
```

```python
transf
```

```python
img_camera_transl = transform.warp(img_camera, transf.inverse)
imshow(img_camera_transl)
```

```python
theta = np.radians(30)
transf = transform.SimilarityTransform(np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
]))
```

```python
img_camera_rotation = transform.warp(img_camera, transf.inverse)
imshow(img_camera_rotation)
```

```python
h, w = img_camera.shape[:2]
cx = w / 2
cy = h / 2
theta = np.radians(30)
```

```python
T1 = np.array([
    [1, 0, -cx],
    [0, 1, -cy],
    [0, 0, 1]
])

R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
])

T2 = np.array([
    [1, 0, cx],
    [0, 1, cy],
    [0, 0, 1]
])
```

```python
M = T2 @ R @ T1
```

```python
trans1 = transform.SimilarityTransform(M)
img_camera_rotation_center = transform.warp(img_camera, trans1.inverse)
```

```python
imshow(img_camera_rotation_center)
```

```python
scale = transform.SimilarityTransform( scale =2)
scaled_img = transform.warp(image = img_camera , inverse_map = scale.inverse)
imshow(scaled_img)
```

```python
center = np.array(img_camera.shape)//2
center
```

```python
# scale around center
shifted = transform.SimilarityTransform(translation= (-center[1] , -center[0] ))
scale = transform.SimilarityTransform(scale=2)
shift_inv = shifted.inverse
```

```python
trans2 = shifted +scale + shift_inv

scale_around_center = transform.warp(image = img_camera , inverse_map = trans2.inverse)
imshow(scale_around_center)
```

```python
# scale around any point
center = [150 ,300]

shifted2 = transform.SimilarityTransform(translation= (-center[1] , -center[0] ))
scale2 = transform.SimilarityTransform(scale=2)
shift_inv2 = shifted.inverse

trans3 = shifted2 +scale2 + shift_inv2

scale_around_center = transform.warp(image = img_camera , inverse_map = trans3.inverse)
imshow(scale_around_center)
```
